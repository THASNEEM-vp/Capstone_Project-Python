**PROJECT OVERVIEW**

The Automobile Market Analytics Dataset is a dataset designed to simulate realistic used car listings. It contains 5,500 vehicle records with detailed information about automobile specifications, ownership history, maintenance records, geographical location, and estimated selling prices.

The dataset incorporates realistic relationships between vehicle age, mileage, technical specifications, condition, and market value, making it suitable for exploratory data analysis and automotive research.



**KEY CHARACTERISTICS**

1. Vehicle depreciation based on age and mileage
2. Engine specifications correlated with body type and fuel type
Ownership history affecting resale value
3. Accident history reducing market price
4. Service records influencing vehicle valuation
5. Fuel efficiency estimated from engine characteristics
6. Random market noise to simulate pricing variability

**PROJECT OBJECTIVE**

Through this project,the main objective is solving a real-world business and economic problem : Predicting vehicle valuation and uncovering the hidden factors that drive depreciation in the secondary automobile market.
Specifically,the analysis aims to find answers to four core business questions:

1. What are the primary price drivers?

Discovering how much weight technical specifications (Horsepower, Torque, Engine Size) carry versus brand prestige (Make) in determining a car's Selling_Price.

2. How severe is asset depreciation?

Quantifying the exact rate at which a car loses its value based on its age (Year) and usage (Mileage).

3. What is the real cost of wear and tear?

Measuring how much value is wiped out by negative history markers like prior accidents (Accident_History) or a lack of maintenance (Service_History).

4. How do market segments perform?

Finding regional price variations (Location) and seeing how different fuel choices (Petrol, Hybrid, Electric) command distinct market premiums.

**FEATURE ENGINEERING**

It ensures that we build meaningful variables that gives deep diving in to the content

 1. Create column **Vehicle_Age** based on the current year (2026) and drop the old Year column safely.'Current_year' represent the reference year for calculating the age of each vehicle.Here performs a vectorized operation using Pandas. It takes the reference year (2026) and subtracts the manufacturing year of each car (stored in the column 'Year') from it.The calculated age is then saved as a brand-new column named 'Vehicle_Age' in DataFrame df.**It ultimately shows how old the car is relative to 2026**

```python
current_year = 2026

if 'Year' in df.columns:
    df['Vehicle_Age'] = current_year - df['Year']
    df = df.drop(columns=['Year'])
display(df[['Brand','Vehicle_Age']].head())
```

OBSERVATON:Depreciation is the single biggest factor influencing used car prices. A model learns age-based devaluation much more naturally than absolute years.

2. Create **Mileage_Per_Year** (Measures yearly Usage Intensity)

 Suppose there is two cars with both have 50,000 miles, but a 2-year-old car with 50,000 miles has been driven aggressively (heavy commercial or rental use), whereas a 10-year-old car with 50,000 miles is lightly used. **This metric captures the rate of wear and tear**.

 Using np.maximum to prevent division by zero for brand-new cars (Vehicle_Age = 0)


```python
df['Mileage_Per_Year']=df['Mileage']/np.maximum(df['Vehicle_Age'],1)
display(df[['Brand','Mileage','Vehicle_Age','Mileage_Per_Year']].head())
```

OBSERVATION: It separates a lightly driven weekend car from an aggressively driven commercial or rental vehicle, capturing real-world wear and tear.

3. Create  the column **'Is_Luxury_Brand'** (Brand Tier Flag)

 Grouping Brand into a binary category (1 for luxury brands like BMW, Mercedes-Benz, Audi, and 0 for mainstream brands like Toyota, Honda, Ford, etc).
 Luxury cars depreciate differently and command higher base price.Giving the model an explicit tier flag helps it learn brand prestige faster.Here creates a list of strings named luxury_brands containing the names of high-end car brands.The .apply() method is a Pandas function that goes through every single row in the 'Brand' column, one by one, and runs a function on each entry.Lambda function return 1 for luxury cars and 0 for mainstream brands.



```python
luxury_brands=['BMW', 'Mercedes-Benz', 'Audi', 'Lexus', 'Porsche', 'Land Rover', 'Jaguar']
df['Is_Luxury_Brand']=df['Brand'].apply(lambda x: 1 if x in luxury_brands else 0)
display(df[['Brand','Is_Luxury_Brand','Vehicle_Age','Selling_Price']].head())
```

OBSERVATION: Luxury cars follow a completely different pricing and depreciation curve than everyday commuter cars, helping the model instantly recognize brand prestige.

4. Create column **Power_to_Size_Ratio** that measures how efficiently the engine produces power.

```python
df['Power_to_Size_Ratio'] = (df['Horsepower'] / df['Engine_Size']).round(2)
display(df[['Horsepower', 'Engine_Size', 'Power_to_Size_Ratio']].head())
```

OBSERVATION:This is key engineering metric that represents the efficiency and performance intensity of an engine.

 **UDF to categorize car age into life stages(Depreciation Tier)** - A custom function combining age and mileage of cars into functional categories like "Nearly New", "Moderately Used", or "High Mileage ".

 UDFs let's write custom Python functions to map complex rules that standard pandas operations can't easily handle.

```python
def categorize_age(Age):
    if Age<=2:
        output='Nearly New'
    elif Age<=6:
        output='Moderately Used'
    else:
        output='High Mileage'
    return output

df['Age_Category']=df['Vehicle_Age'].apply(categorize_age)

# Check average price by age category
age_group_price=df.groupby('Age_Category')['Selling_Price'].mean().reset_index()
print(" Average Price by Age Category\n")
print(age_group_price.round(2))
```

**OBSERVATION** : As the age of vehicle increases the selling price also decreases which highlights the less demand caused due to wear and tear.

**Engine_Size_Category(UDF)** - A custom function to categorize engine size and labels it as "Compact","Standard",or "Heavy-Duty"

```python
def categorize_engine(Engine_Size):

    if Engine_Size<1.5:
        return 'Compact'
    elif Engine_Size<=3.0:
        return 'Standard'
    else:
        return 'Heavy-Duty'

df['Engine_Category']=df['Engine_Size'].apply(categorize_engine)
print(df[['Engine_Size', 'Engine_Category','Power_to_Size_Ratio','Brand','Model','Selling_Price']].head(5))
```

**OBSERVATION** :Segmented vehicles based on industry-standard displacement thresholds:

Compact(< 1.5L): Economy, fuel-efficient, typically lower-priced vehicles.

Standard(1.5L to 3.0L): The mainstream market baseline for standard sedans, crossovers, and family cars.

Performance(> 3.0L): High-power, sports cars, heavy-duty SUVs, or luxury vehicles commanding a price premium.

 errorbar='ci' (Confidence Interval):It shows the uncertainty or reliability of the average.If the error bar is short, it means the data points for that engine category are tightly grouped around the average.  capsize=0.1:Without a capsize, error bars are just straight vertical lines that can look messy or hard to read. Adding a capsize (like 0.1) puts neat little perpendicular caps on them, which is a standard best practice for making charts look clean, professional, and publication-ready.

**7. Mileage_Per_Year vs. Selling_Price Scatter Plot** - A vital analytical chart showing how vehicle usage intensity (the number of miles driven per year) impacts overall market depreciation and selling price.

```python
# Set professional plotting style
sns.set_theme(style="whitegrid")
plt.figure(figsize=(10, 6))

# Create Scatter Plot with regression trendline for Mileage Per Year vs Selling Price
sns.regplot(
    data=df, x='Mileage_Per_Year',
    y='Selling_Price',
    scatter_kws={'alpha': 0.4,'color':'teal'},
    line_kws={'color':'darkorange','linewidth':2}
)

plt.title('Impact of Annual Mileage Intensity on Selling Price',fontsize=14,fontweight='bold')
plt.xlabel('Mileage Per Year (Miles/Year)',fontsize=12)
plt.ylabel('Selling Price ($)',fontsize=12)

# Display the plot
plt.tight_layout()
plt.show()
```

**OBSERVATION** :Downward Trend (Negative Correlation)-The regression line slopes downward from left to right.This confirms that higher annual mileage intensity accelerates vehicle depreciation,resulting in lower market selling prices.

Heavy clustering of datapoints on the left of the graph indicates that for low annual mileage or garagequeen vehicles have higher selling price.

Similarly a vehicle with medium mileage range falling between 50000 and 15000 have the best  normalised selling price.As annual mileage increase the price drops even lower and reaches at a  saturated lower price.


**8. Correlation Heatmap of Numerical Features** - This chart gives an overview of how all numerical variables (such as price, age, mileage etc)correlate with one another.

annot =True shows the correlation coeffecient numbers inside each grid cell
fmt=.2f round the decimal to two places

```python
# Set professional plotting style
sns.set_theme(style="whitegrid")
plt.figure(figsize=(10,8))

# Compute correlation matrix for numerical columns
corr_matrix=df.select_dtypes(include=['number']).corr()

# Create Heatmap with annotations and diverging colormap

sns.heatmap(corr_matrix, annot=True,
    fmt='.2f',cmap='coolwarm',
    vmin=-1, vmax=1, linewidths=0.5
)

plt.title('Correlation Heatmap of Numerical Features',fontsize=14,fontweight='bold')

# Display the plot
plt.tight_layout()
plt.show()
```

**OBSERVATION** : The color bar on the right ranges from +1.00 (dark red,strong positive correlation) to -1.00 (dark blue,strong negative correlation) with values near 0 indicating no linear relationship.

The diagonal line is 1.00s because every feature perfectly correlates with itself.

Features like vehicle age,total mileage,Power to size ratio shows significant negative correlation.

The heatmap is spotting multicollinearity (when two predictor features are too closely related).The horsepower and torque shares high correlation coeffecient of 0.8 whereas mileage and mileage per year also shares 0.84.
